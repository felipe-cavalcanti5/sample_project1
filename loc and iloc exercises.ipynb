```python
# importing pandas and numpy
import pandas as pd
import numpy as np

# crete a sample dataframe
data = pd.DataFrame({
    'age' :     [ 10, 22, 13, 21, 12, 11, 17],
    'section' : [ 'A', 'B', 'C', 'B', 'B', 'A', 'A'],
    'city' :    [ 'Gurgaon', 'Delhi', 'Mumbai', 'Delhi', 'Mumbai', 'Delhi', 'Mumbai'],
    'gender' :  [ 'M', 'F', 'F', 'M', 'M', 'M', 'F'],
    'favourite_color' : [ 'red', np.nan, 'yellow', np.nan, 'black', 'green', 'red']
})

# view the data
data
```

```python

# select all rows with a condition
print("\n\nData with age greater than 15\n\n")
data.loc[data.age >= 15]
```

```python
# select with multiple conditions
data.loc[(data.age>=12) & (data.gender == 'M')]
```

```python
#Slice
data.loc[1:3]
```

```python
# select few columns with a condition
data.loc[(data.age >=12), ['city', 'gender']]
```

```python
# update a column with condition
data.loc[(data.age >= 12), ['section']] = 'M'
data
```

```python
# In this example, if the value in the column age is greater than 20, 
# then the loc function will update the values in the column section with “S” 
# and the values in the column city with Pune:

# update multiple columns with condition
data.loc[(data.age >= 20), ['section', 'city']] = ['S','Pune']
data
```

```python

# select rows with indexes
data.iloc[[0,2]]
```

```python
# select rows with particular indexes and particular columns
data.iloc[[0,2],[1,3]] # the first pair is the indexes and the second one, the columns
```

```python
# select a range of rows
data.iloc[1:3]
```

```python
# select a range of rows and columns
data.iloc[1:3,2:4]
```
